## 8. Data Cleaning

```python
# Fill missing Rainfall using season median
df["Rainfall_mm"] = df.groupby("Season")["Rainfall_mm"].transform(
    lambda x: x.fillna(x.median())
)

# Fill missing Soil Moisture using season median
df["Soil_Moisture_pct"] = df.groupby("Season")["Soil_Moisture_pct"].transform(
    lambda x: x.fillna(x.median())
)

# Fill missing Yield using crop + season median
df["Yield_Tonnes_Ha"] = df.groupby(["Crop", "Season"])["Yield_Tonnes_Ha"].transform(
    lambda x: x.fillna(x.median())
)
```

## 19. Statistical Analysis

```python
from scipy.stats import f_oneway

kharif = df[df["Season"] == "Kharif"]["Yield_Tonnes_Ha"]
rabi = df[df["Season"] == "Rabi"]["Yield_Tonnes_Ha"]
zaid = df[df["Season"] == "Zaid"]["Yield_Tonnes_Ha"]

test_result = f_oneway(kharif, rabi, zaid)

print("ANOVA result:")
print("F-statistic:", test_result.statistic)
print("p-value:", test_result.pvalue)
```

## 20. Unusual Patterns

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x=df["Yield_Tonnes_Ha"]
)

plt.title("Distribution of Crop Yield")
plt.xlabel("Yield (Tonnes/Ha)")
plt.show()
```

```python
q1 = df["Yield_Tonnes_Ha"].quantile(0.25)
q3 = df["Yield_Tonnes_Ha"].quantile(0.75)

iqr = q3 - q1

low_limit = q1 - 1.5 * iqr
high_limit = q3 + 1.5 * iqr

unusual_yield = df[
    (df["Yield_Tonnes_Ha"] < low_limit) |
    (df["Yield_Tonnes_Ha"] > high_limit)
]

print("Number of unusual yield records:", len(unusual_yield))
```
